A simple ** movie recommendation system** using Python, Pandas, and Scikit-learn.

### Workflow

**User selects a movie** → **Recommendation algorithm compares content** → **Suggested movies are displayed**

## 1. Installing and importing libraries

## 3. Convert movie text into features

 **TF-IDF** converts words into numbers so Scikit-learn can compare the movies.

```python
df['content'] = df['genre'] + ' ' + df['description']

vectorizer = TfidfVectorizer(stop_words='english')
content_matrix = vectorizer.fit_transform(df['content'])

print('Number of movies:', len(df))
print('Text feature matrix shape:', content_matrix.shape)
```

## 4. Build the recommendation algorithm

**Cosine similarity** gives a similarity score between movie descriptions. A higher score means the text is more similar.

```python
similarity_matrix = cosine_similarity(content_matrix)

def recommend_movies(movie_title, number_of_recommendations):
    # Find the row number of the movie chosen by the user.
    movie_index = df.index[df['title'] == movie_title][0]

    # Get similarity scores between the selected movie and every movie.
    scores = list(enumerate(similarity_matrix[movie_index]))
    scores = sorted(scores, key=lambda item: item[1], reverse=True)

    # Remove the selected movie itself and keep the best matches.
    best_matches = [item for item in scores if item[0] != movie_index]
    best_matches = best_matches[:int(number_of_recommendations)]

    recommendations = []
    for index, score in best_matches:
        recommendations.append({
            'Recommended Movie': df.iloc[index]['title'],
            'Genre': df.iloc[index]['genre'],
            'Content Similarity (%)': round(float(score) * 100, 1)
        })

    return pd.DataFrame(recommendations)

# Quick test before opening the interface.
recommend_movies('Interstellar', 5)
```

## 5. User interface with Gradio



```python
demo = gr.Interface(
    fn=recommend_movies,
    inputs=[
        gr.Dropdown(choices=df['title'].tolist(), value='Interstellar', label='Choose a movie'),
        gr.Slider(minimum=3, maximum=10, value=5, step=1, label='Number of recommendations')
    ],
    outputs=gr.Dataframe(label='Suggested movies', wrap=True),
    title='Movie Recommendation System',
    description='Select a movie to find other movies with similar genres and descriptions.'
)
demo.launch(share=True, debug=True)
```

## 6. Explanation

- **Input:** The user selects a movie.
- **Data preparation:** Pandas stores the titles, genres, and descriptions in a DataFrame.
- **Feature extraction:** TF-IDF converts the text into numerical features.
- **Recommendation:** Cosine similarity compares the selected movie with the others and ranks them.
- **Output:** Gradio displays the most similar movies in a table.

**Limitation:**
This is a small demonstration dataset and recommends using movie content only. A production system could use a larger real movie dataset and may combine content-based filtering with user ratings or viewing history.
